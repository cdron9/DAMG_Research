```python
import pandas as pd

holdings = pd.read_excel("Portfolio.xlsx", 
                         sheet_name="DVF Tracker",
                         header=2, ## headers are row 3
                         usecols="C:L", ## ticker through total gain.
                         nrows=17,)

## % of portfolio
holdings["Weight"] = (holdings["Market Value"] / holdings["Market Value"].sum()*100)
holdings[["Ticker", "Shares", "Market Value", "Weight"]]
```

```python
## Sector map and weights

sector_map = {
    "NVDA"  : "Technology",
    "ASTS"  : "Technology",
    "MSFT"  : "Technology",
    "ORCL"  : "Technology",
    "APH"   : "Technology",

    "GOOG"  : "Communication Services",
    "META"  : "Communication Services",

    "RTX"   : "Industrials",

    "COP"   : "Energy",

    "BLK"   : "Financial Services",
    "V"     : "Financial Services",
    "PYPL"  : "Financial Services",

    "BKNG"  : "Consumer Cyclical",
    "NKE"   : "Consumer Cyclical",

    "PG"    : "Consumer Defensive",
    "COST"  : "Consumer Defensive",

    "DSI"   : "ETF",
}

holdings["Sector"] = holdings["Ticker"].map(sector_map)

sector_weights = (
    holdings.groupby("Sector")["Weight"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)
sector_weights
```
